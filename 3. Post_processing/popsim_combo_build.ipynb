```python
import pandas as pd

def combine_population_data(year):
    hhgq_df = pd.read_csv(f'{year}/synthetic_households_gq.csv')
    hh_df = pd.read_csv(f'{year}/synthetic_households.csv')
    pergq_df = pd.read_csv(f'{year}/synthetic_persons_gq.csv')
    per_df = pd.read_csv(f'{year}/synthetic_persons.csv')

    hhgq_df['household_id'] += len(hh_df)
    pergq_df['household_id'] += len(hh_df)

    hhfull_df = pd.concat([hh_df, hhgq_df]).drop(columns=['PUMA'])
    hhfull_df['HHADJINC'] = hhfull_df['HHADJINC'].apply(lambda x: max(x, 0))
    hhfull_df.update(hhfull_df[['HHT', 'HUPAC', "BLD", 'GQ_type']].fillna(0))

    perfull_df = pd.concat([per_df, pergq_df]).drop(columns=['PUMA'])
    perfull_df.update(perfull_df[['ESR', 'COW', 'WKHP', 'SCHG', 'MIL', 'SCHL', 'OCCP', 'WKW']].fillna(0))

    # Save combined data or return it
    hhfull_df.to_csv(f'../abm_output/{year}/combined_households.csv', index=False)
    perfull_df.to_csv(f'../abm_output/{year}/combined_persons.csv', index=False)

# Example usage:
combine_population_data('2022')
```

# Merging HH and GQ populations

## Reading syntheic data

```python
hhgq_df = pd.read_csv(fr'{year}\synthetic_households_gq.csv')
hh_df = pd.read_csv(fr'{year}\synthetic_households.csv')
```

```python
# reading GQ syn hh as it has Type
# pergq_fname = os.path.join(runFolder_GQ, 'synthetic_persons_gq.csv')
# per_fname = os.path.join(runFolder, 'synthetic_persons.csv')
pergq_df = pd.read_csv(fr'{year}\synthetic_persons_gq.csv')
per_df = pd.read_csv(fr'{year}\synthetic_persons.csv')

hhgq_df['household_id'] = len(hh_df) + hhgq_df['household_id']
pergq_df['household_id'] = len(hh_df) + pergq_df['household_id']
```

```python
hhfull_df = pd.concat([hh_df, hhgq_df])
hhfull_df = hhfull_df[[item for item in hhfull_df.columns if item not in ['PUMA']]]
```

```python
hhfull_df['HHADJINC'] = hhfull_df['HHADJINC'].apply(lambda x : x if x > 0 else 0)
```

```python
perfull_df = pd.concat([per_df, pergq_df])
perfull_df = perfull_df[[item for item in perfull_df.columns if item not in ['PUMA']]]
```

# Fixing null values 

```python
#na cols
colsnahh = ['HHT', 'HUPAC', "BLD", 'GQ_type']
for col in colsnahh:
    hhfull_df[col] = hhfull_df[col].fillna(0)
```

```python
#perfull_df[ESR,COW, WKHP, SCHG, MIL, SCHL, OCCP, WKW]
colsnaper = ['ESR', 'COW', 'WKHP', 'SCHG', 'MIL', 'SCHL', 'OCCP', 'WKW']
for col in colsnaper:
    perfull_df[col] = perfull_df[col].fillna(0)
```
